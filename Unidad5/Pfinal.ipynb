```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from scipy import stats


df1 = pd.read_csv("grupo1.csv")
df2 = pd.read_csv("grupo2.csv")
df3 = pd.read_csv("grupo3.csv")



print("GRUPO 1 - Primeras filas:")
print(df1.head())

print("\nGRUPO 2 - Primeras filas:")
print(df2.head())

print("\nGRUPO 3 - Primeras filas:")
print(df3.head())



print("\n")
print("ESTADISTICAS DESCRIPTIVAS - GRUPO 1:")
print(df1.describe())

print("\nESTADISTICAS DESCRIPTIVAS - GRUPO 2:")
print(df2.describe())

print("\nESTADISTICAS DESCRIPTIVAS - GRUPO 3:")
print(df3.describe())




grupos  = [df1, df2, df3]
nombres = ["Grupo 1", "Grupo 2", "Grupo 3"]
colores = ["#4C72B0", "#DD8452", "#55A868"]
colores_linea = ["#1a3a6b", "#a0522d", "#1f6b3a"]

for df, nombre, color, color_l in zip(grupos, nombres, colores, colores_linea):

    print("\n")
    print(f"  ANÁLISIS: {nombre}")
    print("\n")

    X = df["HorasEstudio"].values.reshape(-1, 1)
    y = df["Calificacion"].values


    modelo = LinearRegression()
    modelo.fit(X, y)

    pendiente  = modelo.coef_[0]
    intercepto = modelo.intercept_
    r2         = modelo.score(X, y)

    print(f"\nEcuación de regresión:")
    print(f"  Calificación = {pendiente:.4f} × HorasEstudio + {intercepto:.4f}")
    print(f"\nCoeficiente de determinación R² = {r2:.4f}")


    r_pearson, p_valor = stats.pearsonr(df["HorasEstudio"], df["Calificacion"])

    print(f"\nCorrelación de Pearson:")
    print(f"  r = {r_pearson:.4f}")
    print(f"  p-valor = {p_valor:.6f}")


    if abs(r_pearson) >= 0.8:
        fuerza = "fuerte"
    elif abs(r_pearson) >= 0.5:
        fuerza = "moderada"
    else:
        fuerza = "débil"

    direccion = "positiva" if r_pearson > 0 else "negativa"
    print(f"  → Correlación {fuerza} {direccion}")


    plt.figure(figsize=(7, 4))
    plt.hist(df["Calificacion"], bins=10, color=color, edgecolor="white", alpha=0.85)
    plt.axvline(df["Calificacion"].mean(), color=color_l, linestyle="--",
                linewidth=2, label=f"Media: {df['Calificacion'].mean():.1f}")
    plt.title(f"{nombre} — Histograma de Calificaciones", fontweight="bold")
    plt.xlabel("Calificación Final")
    plt.ylabel("Frecuencia")
    plt.legend()
    plt.tight_layout()
    plt.savefig(f"histograma_{nombre.replace(' ', '')}.png", dpi=120)
    plt.show()

    x_line = np.linspace(df["HorasEstudio"].min(), df["HorasEstudio"].max(), 200)
    y_line = pendiente * x_line + intercepto

    plt.figure(figsize=(7, 4))
    plt.scatter(df["HorasEstudio"], df["Calificacion"],
                color=color, alpha=0.7, edgecolors="white", s=70, label="Estudiantes")
    plt.plot(x_line, y_line, color=color_l, linewidth=2.5,
             label=f"y = {pendiente:.2f}x + {intercepto:.2f}")
    plt.title(f"{nombre} — Dispersión y Línea de Regresión", fontweight="bold")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.legend()


    texto = f"r = {r_pearson:.4f}\nR² = {r2:.4f}"
    plt.gca().text(0.05, 0.95, texto, transform=plt.gca().transAxes,
                   fontsize=9, verticalalignment="top",
                   bbox=dict(boxstyle="round", facecolor="white", alpha=0.7))
    plt.tight_layout()
    plt.savefig(f"regresion_{nombre.replace(' ', '')}.png", dpi=120)
    plt.show()



print("\n")
print("  RESUMEN FINAL ")
print("\n")

for df, nombre in zip(grupos, nombres):
    X = df["HorasEstudio"].values.reshape(-1, 1)
    y = df["Calificacion"].values
    modelo = LinearRegression()
    modelo.fit(X, y)
    r_pearson, _ = stats.pearsonr(df["HorasEstudio"], df["Calificacion"])
    print(f"\n{nombre}:")
    print(f"  y = {modelo.coef_[0]:.4f}x + {modelo.intercept_:.4f}")
    print(f"  r = {r_pearson:.4f}   |   R² = {modelo.score(X, y):.4f}")
```
