```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

premios = [0, 10, 20, 50]
probabilidades = [0.50, 0.30, 0.15, 0.05]

suma_p = sum(probabilidades)
print(f"Suma de probabilidades: {suma_p}")

esperanza = sum(x * p for x, p in zip(premios, probabilidades))
print(f"Valor esperado: {esperanza}")

varianza = sum(((x - esperanza)**2) * p for x, p in zip(premios, probabilidades))
desviacion = varianza**0.5
print(f"Varianza: {varianza}")
print(f"Desviación estándar: {desviacion}")

acumulada = np.cumsum(probabilidades)
print(f"Distribución acumulada: {acumulada}")

plt.step(premios, acumulada, where='post')
plt.grid(True)
plt.show()

media = 70
desv_std = 4
num_datos = 500

datos = np.random.normal(media, desv_std, num_datos)
x_rango = np.linspace(media - 4*desv_std, media + 4*desv_std, 100)
pdf = norm.pdf(x_rango, media, desv_std)

plt.figure(figsize=(10, 5))
plt.hist(datos, bins=15, density=True, alpha=0.4, color='b')
plt.plot(x_rango, pdf, 'k', linewidth=2)
plt.grid(True)
plt.show()
```
